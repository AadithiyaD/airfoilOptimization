```python
"""
Generate a PARSEC airfoil, write .dat file, and plot the coordinates.
    Modify PARSEC_PARAMS below to test different airfoil shapes.
"""

import os
from pathlib import Path
import numpy as np
import plotly.graph_objects as go
from src.airfoilTools import Airfoil, parsecParams
```

```python
# ===== USER INPUT: Specify PARSEC parameters here =====
PARSEC_PARAMS = parsecParams(
    r_le=0.012,
    X_up=0.242,
    Z_up=0.048,
    Z_XXup=-0.4,
    
    X_lo=0.242,
    Z_lo=-0.048,
    Z_XXlo=0.4,
    
    Z_te=0,
    delta_Z_te=0,
    alpha_te=-2.23,
    beta_te=7.40,
)
```

```python
# Create data/temp directory if it doesn't exist
Path('data/temp').mkdir(parents=True, exist_ok=True)

# Generate airfoil with unique process ID (for temp file naming)
pid = os.getpid()
foil = Airfoil(name='test_foil', pid=pid, params=PARSEC_PARAMS)

# Write .dat file
output_dat = Path('data/temp/test_foil_output.dat')
foil.write_dat_file(out_path=output_dat)
```

```python
# Load and plot the coordinates
coords = np.genfromtxt(output_dat, skip_header=1)
x_coords = coords[:, 0]
y_coords = coords[:, 1]

mshd_coords = np.genfromtxt("data/temp/e423.dat", skip_header=1)
mshd_x_coords = mshd_coords[:, 0]
mshd_y_coords = mshd_coords[:, 1]

# Create plot
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x_coords,
    y=y_coords,
    mode='lines+markers',
    name='Airfoil Shape',
    line=dict(color='blue', width=2),
    marker=dict(size=4)
))

fig.add_trace(go.Scatter(
    x=mshd_x_coords,
    y=mshd_y_coords,
    mode='lines+markers',
    name='MSHD Airfoil',
    line=dict(color='black', width=2),
    marker=dict(size=4)
))

fig.update_layout(
    title='PARSEC Airfoil',
    xaxis_title='x/c',
    yaxis_title='y/c',
    template='plotly',
    hovermode='closest',
    width=1000,
    height=600
)

fig.update_yaxes(range=[-0.2, 0.2])

fig.show()
```
