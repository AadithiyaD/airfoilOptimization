Use this file to plot an airfoil with corresponding PARSEC parameters and to write it out as a .dat file.
You can also compare the generated airfoil to a reference airfoil.

As a sample, NACA0012 can be represented with the following parameters - 
```python
PARSEC_PARAMS = parsecParams(
    r_le=0.0144,X_up=0.3,Z_up=0.06,Z_XXup=-0.41,
    X_lo=0.3,Z_lo=-0.06,Z_XXlo=0.41,
    Z_te=0,delta_Z_te=0.00256,alpha_te=0,beta_te=12.5,)
```

Parameters for different airfoils can be found through a least squares fit.

````python
import os
from pathlib import Path
import numpy as np
import plotly.graph_objects as go
from src.airfoilTools import Airfoil, parsecParams
````

````python
# ===== USER INPUT: Specify PARSEC parameters here =====
PARSEC_PARAMS = parsecParams(
    r_le=0.008,
    X_up=0.25,
    Z_up=0.056,
    Z_XXup=-0.11,
    
    X_lo=0.5,
    Z_lo=-0.05,
    Z_XXlo=0.4,
    
    Z_te=0,
    delta_Z_te=0.007,
    alpha_te=0.4,
    beta_te=-0.18,
)
````

````python
# Create data/temp directory if it doesn't exist
Path('data/temp').mkdir(parents=True, exist_ok=True)

# Generate airfoil with unique process ID (for temp file naming)
pid = os.getpid()
foil = Airfoil(name='test_foil', pid=pid, params=PARSEC_PARAMS)

# Write .dat file
output_dat = Path('data/temp/test_foil_output.dat')
foil.write_dat_file(out_path=output_dat)
````

````python
# Load and plot the coordinates
coords = np.genfromtxt(output_dat, skip_header=1)
x_coords = coords[:, 0]
y_coords = coords[:, 1]

ref_coords = np.genfromtxt("data/temp/naca0012.dat", skip_header=1)
ref_x_coords = ref_coords[:, 0]
ref_y_coords = ref_coords[:, 1]

# Create plot
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=x_coords,
    y=y_coords,
    mode='lines+markers',
    name='Airfoil Shape',
    line=dict(color='blue', width=2),
    marker=dict(size=4)
))

fig.add_trace(go.Scatter(
    x=ref_x_coords,
    y=ref_y_coords,
    mode='lines+markers',
    name='Reference Airfoil',
    line=dict(color='black', width=2),
    marker=dict(size=4)
))

fig.update_layout(
    title='PARSEC Airfoil',
    xaxis_title='x/c',
    yaxis_title='y/c',
    template='plotly',
    hovermode='closest',
    width=1000,
    height=600
)

fig.update_yaxes(range=[-0.2, 0.2])

fig.show()
````
